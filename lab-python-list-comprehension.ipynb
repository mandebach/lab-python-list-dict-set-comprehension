````python
# Step 1

def initialize_inventory(products):
    inventory = {product: int(input(f"Enter the quantity of {product}: ")) for product in products}
    return inventory
````

````python
# Step 2: Modify the function get_customer_orders so it prompts the user to enter the number of customer orders and 
# gathers the product names using a loop and user input. Use comprehension.


def get_customer_orders():
    # Prompt the user for the total number of orders
    num_order = int(input("How many customer orders would you like to enter?: "))
    
    # Set comprehension to get product names using a loop
    customer_orders = {
        input("Enter the name of the product you would like to order: ").strip().lower()
        for i in range(num_order)
    }
              

    return customer_orders
````

````python
# Step 3: Add a new function to calculate the total price of the customer order. 
# For each product in customer_orders, prompt the user to enter the price of that product. 
# Use comprehension to calculate the total price. Note: assume that the user can only have 1 unit of each product.

def total_price(customer_order):
    # For each item in the customer_order list, ask for the price and add it 
    customer_total = sum(float(input(f"Enter the price for the {item}: ")) for item in customer_order)
    print(f"The total price is ${customer_total:.2f}")
    return customer_total
````

````python
# Step 4: Modify the update_inventory function to remove the product from the inventory if its quantity becomes zero 
# after fulfilling the customer orders. Use comprehension to filter out the products with a quantity of zero from the inventory.

def update_inventory(customer_orders, inventory):
    # 1. Subtract 1 for each ordered item that is in the inventory
    for product in customer_orders:
        if product in inventory:
            inventory[product] -= 1

    # 2. Dictionary comprehension: filter out products with a quantity of 0 (or less)
    inventory = {product: quantity for product, quantity in inventory.items() if quantity > 0}
    return inventory
````

````python
def calculate_order_statistics(customer_orders, products):
    # Total unique products ordered is the size of the set
    total_products_ordered = len(customer_orders)
    
    # Percentage of total product types ordered
    percentage_ordered = (total_products_ordered / len(products)) * 100
    
    # Return both values (as a tuple)
    return total_products_ordered, percentage_ordered
````

````python
def print_order_statistics(order_statistics):
    print("Order Statistics:")
    print(f"Total Products Ordered: {order_statistics[0]}")
    print(f"Percentage of Products Ordered: {order_statistics[1]:.1f}%")
````

````python
def print_updated_inventory(inventory):
    print("\nUpdated Inventory:")
    for product, quantity in inventory.items():
        print(f"{product}: {quantity}")
````

````python
# Define the initial list of available products
products = ["t-shirt", "mug", "hat", "book", "keychain"]

# 1. Initialize the inventory with user-entered quantities
inventory = initialize_inventory(products)

# 2. Get the set of customer orders from the user
customer_orders = get_customer_orders()

# 3. Calculate order statistics
order_stats = calculate_order_statistics(customer_orders, products)

# 4. Print the order statistics
print_order_statistics(order_stats)

# 5. Update the inventory based on customer orders
inventory = update_inventory(customer_orders, inventory)

# 6. Print the updated inventory
print_updated_inventory(inventory)
````

````python
# Print the total price of the customer order.

# Call the function and store the result
order_total = total_price(customer_orders)

# Print the final result (if not already printed inside the function)
print(f"Total Price of Order: ${order_total:.2f}")
````
